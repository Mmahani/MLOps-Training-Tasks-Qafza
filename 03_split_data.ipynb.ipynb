## Reading Data



```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split

PROJECT_DIR = Path.cwd()
ARTIFACTS = PROJECT_DIR / "artifacts"

labeled = pd.read_csv(ARTIFACTS / "02_labeled.csv")
labeled["order_purchase_timestamp"] = pd.to_datetime(
    labeled["order_purchase_timestamp"], errors="coerce"
)

print("Start date:", labeled["order_purchase_timestamp"].min())
print("End date:", labeled["order_purchase_timestamp"].max())
```

##  Splitting (Data Splitting)


```python
train_val, test = train_test_split(
    labeled,
    test_size=0.20,
    random_state=42,
    stratify=labeled["is_late"]
)

train, validation = train_test_split(
    train_val,
    test_size=0.25,
    random_state=42,
    stratify=train_val["is_late"]
)
```

## Checking Sizes and Lag Ratio


```python
for name, df in {
    "train": train,
    "validation": validation,
    "test": test
}.items():
    print(
        f"{name}: rows={len(df):,}, late_ratio={df['is_late'].mean():.4f}"
    )
```

## Saving Files in CSV Format

```python
train.to_csv(ARTIFACTS / "03_train.csv", index=False, encoding="utf-8-sig")
validation.to_csv(ARTIFACTS / "03_validation.csv", index=False, encoding="utf-8-sig")
test.to_csv(ARTIFACTS / "03_test.csv", index=False, encoding="utf-8-sig")

print("Train, validation, and test files saved")
```
